```python
# Import required libraries
import asyncio
import time
from autogen_agentchat.agents import AssistantAgent
from autogen_agentchat.messages import TextMessage
from autogen_core import CancellationToken
from autogen_ext.models.ollama import OllamaChatCompletionClient

print('✅ All libraries imported successfully!')
```

```python
# Check available ULTRA-TINY Llama models
print('🔍 Checking available ULTRA-TINY Llama models...')
print('\nAvailable ULTRA-TINY models (fastest possible):')
print('- qwen2.5:0.5b (ultra-tiny, lightning fast - 397MB)')
print('- tinyllama:1.1b (super tiny, very fast - 637MB)')
print('- gemma2:2b (tiny, fast - 1.6GB)')
print('\n💡 For this demo, we will use ONLY the tiniest models for maximum speed!')
```

```python
# Create ONLY the tiniest, fastest models
print('🦙 Creating ULTRA-TINY Llama models for maximum speed...')

# Model 1: Ultra-tiny, lightning fast model for instant responses
ultra_tiny_model = OllamaChatCompletionClient(
    model='qwen2.5:0.5b'  # Ultra-tiny, only 397MB model
)

# Model 2: Super tiny, very fast model for quick responses
super_tiny_model = OllamaChatCompletionClient(
    model='tinyllama:1.1b'  # Super tiny, only 637MB model
)

# Model 3: Tiny, fast model for quality responses
tiny_model = OllamaChatCompletionClient(
    model='gemma2:2b'  # Tiny, only 1.6GB model
)

print('✅ Three ULTRA-TINY models created for maximum speed!')
print('   - qwen2.5:0.5b (ultra-tiny, lightning fast - 397MB)')
print('   - tinyllama:1.1b (super tiny, very fast - 637MB)')
print('   - gemma2:2b (tiny, fast - 1.6GB)')
```

```python
# Create specialized agents with ULTRA-TINY models
print('🤖 Creating specialized agents with ULTRA-TINY models...')

# Agent 1: Ultra-Tiny Responder (using ultra-tiny model)
ultra_tiny_responder = AssistantAgent(
    name='UltraTinyResponder',
    system_message='You are an ultra-tiny, lightning-fast assistant. Provide instant, concise answers.',
    model_client=ultra_tiny_model
)

# Agent 2: Super Tiny Assistant (using super tiny model)
super_tiny_assistant = AssistantAgent(
    name='SuperTinyAssistant',
    system_message='You are a super tiny, very fast assistant. Provide quick, good quality answers.',
    model_client=super_tiny_model
)

# Agent 3: Tiny Expert (using tiny model)
tiny_expert = AssistantAgent(
    name='TinyExpert',
    system_message='You are a tiny but efficient expert. Provide fast, quality responses.',
    model_client=tiny_model
)

print('✅ Three ULTRA-TINY agents created for maximum speed!')
```

```python
# Test 1: Ultra-tiny response with ultra-tiny model
print('⚡ Testing Ultra-Tiny Responder (qwen2.5:0.5b - Ultra-Tiny Model)...')
start_time = time.time()

ultra_tiny_prompt = TextMessage(
    content='What is 2 + 2?', 
    source='user'
)

ultra_tiny_response = await ultra_tiny_responder.on_messages(
    [ultra_tiny_prompt], 
    cancellation_token=CancellationToken()
)

ultra_tiny_time = time.time() - start_time
print(f'⏱️  Response time: {ultra_tiny_time:.2f} seconds')
print(f'💬 Response: {ultra_tiny_response.chat_message.content}')
```

```python
# Test 2: Super tiny response with super tiny model
print('\n⚖️  Testing Super Tiny Assistant (tinyllama:1.1b - Super Tiny Model)...')
start_time = time.time()

super_tiny_prompt = TextMessage(
    content='Explain the concept of recursion in programming.', 
    source='user'
)

super_tiny_response = await super_tiny_assistant.on_messages(
    [super_tiny_prompt], 
    cancellation_token=CancellationToken()
)

super_tiny_time = time.time() - start_time
print(f'⏱️  Response time: {super_tiny_time:.2f} seconds')
print(f'💬 Response: {super_tiny_response.chat_message.content[:200]}...')
```

```python
# Test 3: Tiny response with tiny model
print('\n🎯 Testing Tiny Expert (gemma2:2b - Tiny Model)...')
start_time = time.time()

tiny_prompt = TextMessage(
    content='What is a function in programming?', 
    source='user'
)

tiny_response = await tiny_expert.on_messages(
    [tiny_prompt], 
    cancellation_token=CancellationToken()
)

tiny_time = time.time() - start_time
print(f'⏱️  Response time: {tiny_time:.2f} seconds')
print(f'💬 Response: {tiny_response.chat_message.content[:200]}...')
```

```python
# Compare performance and demonstrate ULTRA-TINY multi-model benefits
print('\n📊 ULTRA-TINY Multi-Model Performance Comparison')
print('=' * 60)
print(f'⚡ Ultra-Tiny Responder (qwen2.5:0.5b): {ultra_tiny_time:.2f}s')
print(f'⚖️  Super Tiny Assistant (tinyllama:1.1b):   {super_tiny_time:.2f}s')
print(f'🎯 Tiny Expert (gemma2:2b):            {tiny_time:.2f}s')
print('\n📈 Performance Analysis:')
print(f'   - Fastest: qwen2.5:0.5b ({ultra_tiny_time:.2f}s)')
print(f'   - Slowest: gemma2:2b ({tiny_time:.2f}s)')
print(f'   - Speed difference: {tiny_time - ultra_tiny_time:.2f}s')
print(f'   - qwen2.5:0.5b is {tiny_time/ultra_tiny_time:.1f}x faster than gemma2:2b')

print('\n🎯 Key Benefits of ULTRA-TINY Multi-Model Setup:')
print('1. **Lightning-fast responses** on any hardware')
print('2. **Minimal resource usage** with ultra-tiny models')
print('3. **Instant testing** without any waits')
print('4. **Efficient learning** with immediate feedback')
print('5. **No system slowdown** or crashes')
```
