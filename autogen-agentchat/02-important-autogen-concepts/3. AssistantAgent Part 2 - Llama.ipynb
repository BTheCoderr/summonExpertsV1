## Advanced AssistantAgent Features with Llama

This notebook shows advanced features of AssistantAgent using your local Llama model.

```python
from autogen_ext.models.ollama import OllamaChatCompletionClient
from autogen_agentchat.agents import AssistantAgent
from autogen_agentchat.messages import TextMessage
from autogen_core import CancellationToken

llama_client = OllamaChatCompletionClient(
    model='llama3.2:3b'
)

print('✅ Llama client ready!')
```

```python
coding_expert = AssistantAgent(
    name='CodingExpert',
    system_message='You are an expert Python programmer. Provide clear, well-documented code examples.',
    model_client=llama_client
)

math_tutor = AssistantAgent(
    name='MathTutor',
    system_message='You are a patient math tutor. Explain concepts step by step.',
    model_client=llama_client
)

print('🤖 Specialized agents created!')
```

```python
code_question = TextMessage(
    content='Write a Python function to calculate fibonacci numbers with memoization.',
    source='user'
)

code_response = await coding_expert.on_messages([code_question], cancellation_token=CancellationToken())
print('✅ Code explanation successful!')
print(f'Explanation: {code_response.chat_message.content[:200]}...')
```

```python
math_question = TextMessage(
    content='Explain how to solve quadratic equations step by step.',
    source='user'
)

math_response = await math_tutor.on_messages([math_question], cancellation_token=CancellationToken())
print('✅ Code optimization successful!')
print(f'Optimization: {math_response.chat_message.content[:200]}...')
```

```python
same_question = TextMessage(
    content='What is recursion?',
    source='user'
)

coding_response = await coding_expert.on_messages([same_question], cancellation_token=CancellationToken())
math_response = await math_tutor.on_messages([same_question], cancellation_token=CancellationToken())

print('🤖 Coding Expert on Recursion:')
print(coding_response.chat_message.content)
print('\n' + '='*50 + '\n')
print('🧮 Math Tutor on Recursion:')
print(math_response.chat_message.content)
```
