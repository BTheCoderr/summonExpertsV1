## Streaming Messages with Llama

This notebook demonstrates streaming message capabilities with your local Llama model.

```python
from autogen_ext.models.ollama import OllamaChatCompletionClient
from autogen_agentchat.agents import AssistantAgent
from autogen_agentchat.messages import TextMessage
from autogen_core import CancellationToken
import asyncio

print('✅ All modules imported successfully!')
```

```python
llama_client = OllamaChatCompletionClient(
    model='llama3.2:3b'
)

print('🦙 Llama client ready!')
```

```python
streaming_agent = AssistantAgent(
    name='StreamingAgent',
    system_message='You are a helpful AI assistant. Provide detailed, step-by-step explanations.',
    model_client=llama_client
)

print('🤖 Streaming agent created!')
```

```python
message = TextMessage(
    content='Explain how machine learning works in simple terms.',
    source='user'
)

print('📝 Sending message...')
print('🔄 Response will stream in real-time...')

response = await streaming_agent.on_messages([message], cancellation_token=CancellationToken())
print('\n✅ Complete response received:')
print(response.chat_message.content)
```
