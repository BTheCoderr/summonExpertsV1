# UserProxyAgent - Llama Version

This notebook demonstrates how to use AutoGen UserProxyAgent with Llama.

## What is UserProxyAgent?

UserProxyAgent represents a human user in the conversation. It can:
- Receive messages from other agents
- Process and respond to messages
- Control the flow of conversation
- Simulate human behavior

```python
# Import required modules
from autogen_ext.models.ollama import OllamaChatCompletionClient
from autogen_agentchat.agents import UserProxyAgent, AssistantAgent
from autogen_agentchat.messages import TextMessage
from autogen_core import CancellationToken

print('✅ All modules imported successfully!')
```

```python
# Initialize Llama client
llama_client = OllamaChatCompletionClient(
    model='llama3.2:3b'
)

print('🦙 Llama client ready!')
```

```python
# Create agents with correct modern syntax
assistant = AssistantAgent(
    name='LlamaAssistant',
    system_message='You are a helpful AI assistant. Answer questions clearly and concisely.',
    model_client=llama_client
)

# UserProxyAgent with correct parameters
user_proxy = UserProxyAgent(
    name='UserProxy',
    description='A human user who asks questions and receives answers'
)

print('🤖 Both agents created successfully!')
```

```python
# Test basic agent communication
print('🧪 Testing basic agent communication...')

# Create a simple message
message = TextMessage(content='Hello! Can you help me with a math problem?', source='user')

# Send message to assistant
response = await assistant.on_messages(
    messages=[message],
    cancellation_token=CancellationToken()
)

print('💬 Conversation started!')
print('📝 Initial message:', message.content)
print('🤖 Assistant response:', response.chat_message.content)
```

```python
# Test UserProxyAgent functionality
print('🧪 Testing UserProxyAgent...')

# Create a response message from user proxy
user_response = TextMessage(
    content='Thank you! Can you also explain how you solved it?', 
    source='UserProxy'
)

print('👤 UserProxy generated response:', user_response.content)
print('✅ UserProxyAgent is working correctly!')
```
