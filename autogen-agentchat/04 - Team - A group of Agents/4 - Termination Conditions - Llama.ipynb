When you have **3 different models**, you get:

- **Role specialization** - each model optimized for its purpose
- **Better problem-solving** - multiple perspectives on the same issue

- **True AI team collaboration** - like having 3 different experts working together

### Question: What is a termination condition?
Answer: When agents are running in a team, a Termination Condition defines the criteria for halting a conversation or process. 

It is crucial for ensuring that runs are terminated at the right time, rather than running endlessly.

## 🎯 Our 3-Model Team Strategy:

1. **Creative Agent (qwen2.5:0.5b)**: Generates ideas and solutions
2. **Critic Agent (tinyllama:1.1b)**: Analyzes and provides feedback
3. **Quality Agent (gemma2:2b)**: Refines and improves the final output

This creates a **complete workflow** where each model contributes its unique strengths!

```python
# Import required libraries for ultimate 3-model Llama team
from autogen_agentchat.agents import AssistantAgent
from autogen_agentchat.conditions import MaxMessageTermination, TextMentionTermination
from autogen_agentchat.teams import RoundRobinGroupChat
from autogen_agentchat.ui import Console
from autogen_ext.models.ollama import OllamaChatCompletionClient

print('✅ All libraries imported successfully!')
print('🚀 Ready to create the ULTIMATE 3-model Llama team!')
print('🎯 This will be the most diverse and collaborative AI team possible!')
```

```python
# Create THREE different ultra-tiny Llama model clients for maximum diversity!
print('🦙 Creating THREE different ultra-tiny Llama model clients...')
print('🎯 Each model has different strengths and thinking styles!')

# Model 1: Ultra-tiny, lightning fast for creative responses (397MB)
creative_model = OllamaChatCompletionClient(
    model='qwen2.5:0.5b'  # Ultra-tiny, only 397MB model
)

# Model 2: Super tiny, fast for critical analysis (637MB)
critic_model = OllamaChatCompletionClient(
    model='tinyllama:1.1b'  # Super tiny, only 637MB model
)

# Model 3: Tiny, fast for quality responses (1.6GB)
quality_model = OllamaChatCompletionClient(
    model='gemma2:2b'  # Tiny, only 1.6GB model
)

print('✅ THREE different ultra-tiny Llama models created!')
print('   - qwen2.5:0.5b (397MB) - Lightning fast for creative responses')
print('   - tinyllama:1.1b (637MB) - Fast for critical analysis')
print('   - gemma2:2b (1.6GB) - Fast for quality responses')
print('\n🎯 Key Benefit: 3 different models = 3 different thinking styles!')
print('🚀 This is the ULTIMATE multi-model setup!')
```

```python
# Create the ULTIMATE 3-agent team with DIFFERENT ultra-tiny Llama models
print('🤖 Creating the ULTIMATE 3-agent team with DIFFERENT models...')
print('🎯 Each agent will think differently and contribute unique perspectives!')

# Agent 1: Creative Agent using qwen2.5:0.5b (397MB) - for imaginative responses
creative_agent = AssistantAgent(
    name="CreativeGenius",
    model_client=creative_model,  # Uses qwen2.5:0.5b
    system_message="You are a creative genius using qwen2.5:0.5b. Generate innovative, out-of-the-box ideas and solutions. Think creatively and suggest unique approaches. Your strength is imagination and originality."
)

# Agent 2: Critic Agent using tinyllama:1.1b (637MB) - for critical analysis
critic_agent = AssistantAgent(
    name="CriticalAnalyst",
    model_client=critic_model,  # Uses tinyllama:1.1b
    system_message="You are a critical analyst using tinyllama:1.1b. Provide BRIEF, focused analysis (2-3 sentences max). Identify 1-2 key issues or improvements. ALWAYS end your response with 'ANALYSIS_COMPLETE'. Be concise and direct."
)

# Agent 3: Quality Agent using gemma2:2b (1.6GB) - for refinement and quality
quality_agent = AssistantAgent(
    name="QualityRefiner",
    model_client=quality_model,  # Uses gemma2:2b
    system_message="You are a quality refiner using gemma2:2b. Take the ideas and feedback, then provide a BRIEF, polished solution (2-3 sentences max). ALWAYS end your response with 'REFINEMENT_COMPLETE'. Be concise and focused."
)

print('✅ ULTIMATE 3-agent team created with DIFFERENT models!')
print('   - CreativeGenius: Uses qwen2.5:0.5b (397MB) - for imagination and originality')
print('   - CriticalAnalyst: Uses tinyllama:1.1b (637MB) - for analysis and evaluation')
print('   - QualityRefiner: Uses gemma2:2b (1.6GB) - for synthesis and improvement')
print('\n🎯 This is true multi-model collaboration - each agent thinks differently!')
print('🚀 Watch how they work together to solve problems!')
```

```python
# 🚀 DEMONSTRATION 1: MaxMessageTermination with 3-Model Team
print('🚀 DEMONSTRATION 1: MaxMessageTermination with 3-Model Team')
print('📝 Task: Design a futuristic smart home system')
print('⏱️  Expected time: 5-15 seconds (instead of minutes!)\n')
print('🎯 Termination: Stops after 6 messages maximum (2 from each agent)')
print('\n💡 Watch how the 3 DIFFERENT models collaborate!')
print('   - CreativeGenius: Generates innovative ideas')
print('   - CriticalAnalyst: Provides analysis and feedback')
print('   - QualityRefiner: Refines and improves the solution')

# Create a team with MaxMessageTermination
max_message_team = RoundRobinGroupChat(
    [creative_agent, critic_agent, quality_agent],
    termination_condition=MaxMessageTermination(6)
)

# Run the team and observe the 3 different thinking styles!
await Console(max_message_team.run_stream(task="Design a futuristic smart home system with AI integration."))
```

```python
# 🚀 DEMONSTRATION 2: TextMentionTermination with 3-Model Team
print('🚀 DEMONSTRATION 2: TextMentionTermination with 3-Model Team')
print('📝 Task: Create a sustainable city transportation plan')
print('⏱️  Expected time: 8-20 seconds (instead of minutes!)\n')
print('🎯 Termination: Stops when both critic and quality agents complete their work')
print('\n💡 This shows how different models can have different completion criteria!')
print('   - CreativeGenius: Generates transportation ideas')
print('   - CriticalAnalyst: Analyzes feasibility (says ANALYSIS_COMPLETE)')
print('   - QualityRefiner: Refines the plan (says REFINEMENT_COMPLETE)')

# Create a team with TextMentionTermination for both completion keywords
text_mention_team = RoundRobinGroupChat(
    [creative_agent, critic_agent, quality_agent],
    termination_condition=(
        TextMentionTermination("ANALYSIS_COMPLETE") | 
        TextMentionTermination("REFINEMENT_COMPLETE")
    )
)

# Run the team and watch the 3 models work together!
await Console(text_mention_team.run_stream(task="Create a sustainable city transportation plan with electric vehicles and public transit."))
```

```python
# 🚀 DEMONSTRATION 3: Advanced Combined Termination with 3-Model Team
print('🚀 DEMONSTRATION 3: Advanced Combined Termination with 3-Model Team')
print('📝 Task: Invent a new type of renewable energy technology')
print('⏱️  Expected time: 10-25 seconds (instead of minutes!)\n')
print('🎯 Termination: Stops when both agents complete OR after 9 messages maximum')
print('\n💡 This demonstrates the most advanced termination logic!')
print('   - CreativeGenius: Invents the technology concept')
print('   - CriticalAnalyst: Analyzes technical feasibility')
print('   - QualityRefiner: Refines and documents the invention')

# Create a team with advanced combined termination conditions
advanced_team = RoundRobinGroupChat(
    [creative_agent, critic_agent, quality_agent],
    termination_condition=((TextMentionTermination("ANALYSIS_COMPLETE") | TextMentionTermination("REFINEMENT_COMPLETE")) | MaxMessageTermination(9))
)

# Run the ULTIMATE 3-model team with advanced termination!
await Console(advanced_team.run_stream(task="Invent a new type of renewable energy technology that could power cities."))
```
